## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

```python
### Research question

Can observable content and search-performance signals be used to rank pages by their potential need for review, so that a content team can decide which pages to investigate first?

### Unit of analysis

The unit of analysis will be one content page.

### Output

The intended output is a ranked opportunity score or review queue that helps prioritize pages for human review.

### Decision

The decision is which pages should receive attention first for possible refresh, expansion, protection, pruning, or monitoring.

### Who acts on it

A content or SEO team would use the ranking as decision support when allocating limited review and optimization time.

### Action

A reviewer could inspect the highest-priority pages first and decide whether to refresh, expand, protect, prune, or continue monitoring them.

### Cost of a wrong recommendation

A wrong recommendation could waste limited content resources by prioritizing a page that does not need attention, while a genuinely valuable opportunity could be reviewed later. Therefore, the output should support human judgment rather than automatically determine what action must be taken.
```
