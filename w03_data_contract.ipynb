## 4. Data limits

*What can this data never tell you? Unbalanced history, GSC-only early rows, window overlaps.*

```python
This dataset provides observational historical data, so it can show measured patterns and associations but cannot establish that a particular content change caused an improvement in search performance.

The available history may also be unbalanced across pages and time periods. Some early observations may contain Search Console data without the same breadth of analytics information available later. In addition, overlapping time windows may mean that observations are not fully independent.

Therefore, any scoring or ranking developed from this data should be treated as directional decision support rather than causal proof or a guarantee of future performance.
```
